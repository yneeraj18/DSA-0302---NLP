```python
words = ["connected", "connecting", "connection"]

print("{:<15} {:<15} {:<15} {:<20} {:<15}".format(
    "Word", "Root", "Suffix", "Suffix Type", "Normalized Form"
))
print("-" * 80)

for word in words:
    if word == "connected":
        root = "connect"
        suffix = "ed"
        suffix_type = "Inflectional"
        normalized = "connect"

    elif word == "connecting":
        root = "connect"
        suffix = "ing"
        suffix_type = "Inflectional"
        normalized = "connect"

    elif word == "connection":
        root = "connect"
        suffix = "ion"
        suffix_type = "Derivational"
        normalized = "connect"

    print("{:<15} {:<15} {:<15} {:<20} {:<15}".format(
        word, root, suffix, suffix_type, normalized
    ))
```
