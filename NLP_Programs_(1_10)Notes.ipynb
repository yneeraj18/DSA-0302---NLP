1. Tiny Spell Checker using Word Counts
Algorithm

```python
from collections import Counter
import re

text = """
python is easy to learn
python is a programming language
python is easy and powerful
programming is fun and easy
"""

words = re.findall(r'\w+', text.lower())
word_count = Counter(words)

def edits1(word):
    letters = 'abcdefghijklmnopqrstuvwxyz'

    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]

    deletes = [L + R[1:] for L, R in splits if R]

    replaces = [L + c + R[1:]
                for L, R in splits if R
                for c in letters]

    inserts = [L + c + R
               for L, R in splits
               for c in letters]

    return set(deletes + replaces + inserts)

word = input("Enter a word: ").lower().strip()

if word in word_count:
    print("Spelling is correct")
else:
    suggestions = edits1(word)
    possible = [w for w in suggestions if w in word_count]

    if possible:
        correct = max(possible, key=lambda w: word_count[w])
        print("Spelling is incorrect")
        print("Did you mean:", correct)
        print("Probability/Frequency:", word_count[correct])
    else:
        print("Spelling is incorrect")
        print("No suggestion found")
```

2. Paragraph Analysis

```python
paragraph = input("Enter a paragraph: ")

sentences = paragraph.split(".")

print("\nSentences:")
for sentence in sentences:
    if sentence.strip():
        print(sentence.strip())

words = paragraph.split()

print("\nWords:")
print(words)

print("\nTotal number of words:", len(words))

count = {}

for word in words:
    word = word.lower().strip(".,!?")

    if word in count:
        count[word] += 1
    else:
        count[word] = 1

print("\nWord Frequency:")
for word in count:
    print(word, ":", count[word])

highest = max(count.values())

print("\nMost frequent words:")
for word in count:
    if count[word] == highest:
        print(word, ":", count[word])
```

3. Regular Expression Program

```python
import re

text = """Call me on 9876543210 or 9123456780.
My email is student@gmail.com and teacher@yahoo.com.
The meeting is on 25-12-2026.
Python and NLP are useful Technologies."""

date = re.search(r'\d{2}-\d{2}-\d{4}', text)

if date:
    print("First date:", date.group())

mobiles = re.findall(r'\b[6-9]\d{9}\b', text)

print("\nMobile numbers:")
print(mobiles)

emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)

print("\nEmail IDs:")
print(emails)

capitalized = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)

print("\nCapitalized words:")
print(capitalized)
if re.match(r'call', text, re.IGNORECASE):
    print("\nText starts with 'call'")
else:
    print("\nText does not start with 'call'")
def change_date(match):
    day, month, year = match.group().split("-")
    return year + "/" + month + "/" + day

new_text = re.sub(r'\d{2}-\d{2}-\d{4}', change_date, text)

print("\nText after changing date:")
print(new_text)
hidden_text = re.sub(r'\b[6-9]\d{9}\b', 'XXXXXXXXXX', new_text)

print("\nText after hiding phone numbers:")
print(hidden_text)
```

4. Regular Expression Operators

```python
import re

patterns = {
    "?  Zero or one": r"ab?c",
    "*  Zero or more": r"ab*c",
    "+  One or more": r"ab+c",
    "[] Character set": r"[abc]+",
    "[^] Not in set": r"[^abc]+",
    "^  Starts with": r"^Hello",
    "$  Ends with": r"world$"
}

samples = {
    "?  Zero or one": ["ac", "abc", "abbc"],
    "*  Zero or more": ["ac", "abc", "abbbc"],
    "+  One or more": ["ac", "abc", "abbbc"],
    "[] Character set": ["abc", "cab", "xyz"],
    "[^] Not in set": ["xyz", "hello", "abc"],
    "^  Starts with": ["Hello world", "hello world", "Say Hello"],
    "$  Ends with": ["hello world", "world", "world hello"]
}

for name in patterns:
    print("\n", name)
    print("Pattern:", patterns[name])

    for text in samples[name]:
        if re.search(patterns[name], text):
            print(text, "-> YES")
        else:
            print(text, "-> NO")
```

5. Finite State Automaton — Ends with "ab"

```python
def check_string(string):
    state = "q0"

    print("\nInput:", string)
    print("Path:", state, end=" ")

    for ch in string:

        if state == "q0":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        elif state == "q1":
            if ch == "a":
                state = "q1"
            else:
                state = "q2"

        elif state == "q2":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        print("->", state, end=" ")

    if state == "q2":
        print("\nACCEPTED")
    else:
        print("\nREJECTED")


strings = ["ab", "aab", "abab", "baab", "a", "aba", "abb"]

for string in strings:
    check_string(string)
```

6. Program to Split a Word into Prefix, Root, and Suffix

```python
prefixes = ["un", "re", "dis", "pre", "mis", "over", "under", "non"]
suffixes = ["ing", "ed", "er", "est", "ly", "ness", "ment", "ful", "less", "able", "s"]

def analyze_word(word):
    original_word = word.lower()
    prefix = ""
    suffix = ""
    root = original_word

    for p in sorted(prefixes, key=len, reverse=True):
        if root.startswith(p) and len(root) > len(p) + 2:
            prefix = p
            root = root[len(p):]
            break

    for s in sorted(suffixes, key=len, reverse=True):
        if root.endswith(s) and len(root) > len(s) + 2:
            suffix = s
            root = root[:-len(s)]
            break

    print("Word:", word)
    print("Prefix:", prefix if prefix else "None")
    print("Root:", root)
    print("Suffix:", suffix if suffix else "None")

word = input("Enter a word: ")
analyze_word(word)
```

7. Program to Generate Plural Nouns, Verb Forms, and Comparative and Superlative Adjectives

```python
irregular_nouns = {
    "man": "men",
    "woman": "women",
    "child": "children",
    "foot": "feet",
    "tooth": "teeth",
    "mouse": "mice",
    "person": "people"
}

irregular_verbs = {
    "go": ("went", "going"),
    "eat": ("ate", "eating"),
    "write": ("wrote", "writing"),
    "run": ("ran", "running"),
    "swim": ("swam", "swimming"),
    "have": ("had", "having"),
    "be": ("was/were", "being")
}

irregular_adjectives = {
    "good": ("better", "best"),
    "bad": ("worse", "worst"),
    "far": ("farther", "farthest")
}

def plural(noun):
    if noun in irregular_nouns:
        return irregular_nouns[noun]
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        return noun + "es"
    if noun.endswith("y") and len(noun) > 1 and noun[-2] not in "aeiou":
        return noun[:-1] + "ies"
    if noun.endswith("fe"):
        return noun[:-2] + "ves"
    if noun.endswith("f"):
        return noun[:-1] + "ves"
    return noun + "s"

def verb_forms(verb):
    if verb in irregular_verbs:
        return irregular_verbs[verb]

    if verb.endswith("ie"):
        past = verb[:-2] + "ied"
        ing = verb[:-2] + "ying"
    elif verb.endswith("e") and not verb.endswith("ee"):
        past = verb + "d"
        ing = verb[:-1] + "ing"
    elif (len(verb) >= 3 and
          verb[-1] not in "aeiouwxy" and
          verb[-2] in "aeiou" and
          verb[-3] not in "aeiou"):
        past = verb + verb[-1] + "ed"
        ing = verb + verb[-1] + "ing"
    else:
        past = verb + "ed"
        ing = verb + "ing"

    return past, ing

def adjective_forms(adj):
    if adj in irregular_adjectives:
        return irregular_adjectives[adj]

    if adj.endswith("y"):
        return adj[:-1] + "ier", adj[:-1] + "iest"
    if adj.endswith("e"):
        return adj + "r", adj + "st"
    if len(adj) <= 5:
        return adj + "er", adj + "est"

    return "more " + adj, "most " + adj

noun = input("Enter a noun: ").lower()
verb = input("Enter a verb: ").lower()
adj = input("Enter an adjective: ").lower()

print("\nPlural Noun:", plural(noun))

past, ing = verb_forms(verb)
print("Past Tense:", past)
print("Ing Form:", ing)

comparative, superlative = adjective_forms(adj)
print("Comparative:", comparative)
print("Superlative:", superlative)
```

8. Program to Identify Derivational Suffixes and POS Changes

```python
suffix_data = {
    "ness": ("Adjective", "Noun", "State or quality of"),
    "ment": ("Verb", "Noun", "Result or process of"),
    "tion": ("Verb", "Noun", "Action or process of"),
    "able": ("Verb", "Adjective", "Capable of being"),
    "ful": ("Noun", "Adjective", "Full of"),
    "less": ("Noun", "Adjective", "Without"),
    "ly": ("Adjective", "Adverb", "In a particular manner"),
    "er": ("Verb", "Noun", "A person who performs an action"),
    "ist": ("Noun", "Noun", "A person associated with")
}

def analyze_derivation(word):
    word = word.lower()

    for suffix in sorted(suffix_data, key=len, reverse=True):
        if word.endswith(suffix) and len(word) > len(suffix):
            base = word[:-len(suffix)]
            old_pos, new_pos, meaning = suffix_data[suffix]

            print("Word:", word)
            print("Base:", base)
            print("Suffix:", suffix)
            print("Original Word Class:", old_pos)
            print("New Word Class:", new_pos)
            print("Meaning:", meaning, base)
            return

    print("No known derivational suffix found.")

word = input("Enter a derived word: ")
analyze_derivation(word)
```

9. Program for Morphological Analysis Using NLTK

```python
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

from nltk import word_tokenize, pos_tag
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return 'a'
    elif tag.startswith('V'):
        return 'v'
    elif tag.startswith('N'):
        return 'n'
    elif tag.startswith('R'):
        return 'r'
    return 'n'

def analyze_sentence(sentence):
    words = word_tokenize(sentence)
    tagged_words = pos_tag(words)

    noun_count = 0

    print("\nWord\t\tPOS\t\tLemma\t\tFeature")
    print("-" * 65)

    for word, tag in tagged_words:
        lemma = lemmatizer.lemmatize(word.lower(), get_wordnet_pos(tag))

        if tag.startswith('N'):
            noun_count += 1

        if tag in ('VBD', 'VBN'):
            feature = "Past tense / past participle"
        elif tag in ('VBP', 'VBZ', 'VBG', 'VB'):
            feature = "Present / base / ing form"
        elif tag.startswith('V'):
            feature = "Verb"
        elif tag.startswith('N'):
            feature = "Noun"
        elif tag.startswith('J'):
            feature = "Adjective"
        elif tag.startswith('R'):
            feature = "Adverb"
        else:
            feature = "Other"

        print(f"{word}\t\t{tag}\t\t{lemma}\t\t{feature}")

    print("\nTotal nouns:", noun_count)

sentence = input("Enter a sentence: ")
analyze_sentence(sentence)
```

10. Program to Perform Stemming Using the Porter Stemmer Algorithm

```python
import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing",
    "played",
    "plays",
    "happiness",
    "studies",
    "studying",
    "connected",
    "connection",
    "running",
    "relational"
]

print("Original Word\tStemmed Word")
print("-" * 35)

for word in words:
    stem = stemmer.stem(word)
    print(f"{word}\t\t{stem}")
```
