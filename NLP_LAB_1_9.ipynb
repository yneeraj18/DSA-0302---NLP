1. Regular Expressions – Match and Search Patterns

```python
import re

text = "My email is student@gmail.com and my phone is 9876543210."
email = re.search(r'\w+@\w+\.\w+', text)
phone = re.search(r'\d{10}', text)

if email:
    print("Email found:", email.group())
else:
    print("Email not found")

if phone:
    print("Phone number found:", phone.group())
else:
    print("Phone number not found")
```

2. Finite State Automaton – Strings Ending with ab

```python
def fsa(string):
    state = 0
    for ch in string:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == 'b':
                state = 2
            elif ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0
    return state == 2
s = input("Enter a string: ")
if fsa(s):
    print("ACCEPTED")
else:
    print("REJECTED")
```

3. Morphological Analysis using NLTK

```python
import nltk

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "studies", "cars"]

for word in words:
    print("Word:", word)
    print("Stem:", stemmer.stem(word))
    print("Lemma:", lemmatizer.lemmatize(word))
    print()
```

4. Finite-State Machine for Morphological Parsing – Plural Nouns

```python
def plural(noun):
    if noun.endswith("s"):
        return noun + "es"
    elif noun.endswith("x") or noun.endswith("ch") or noun.endswith("sh"):
        return noun + "es"
    else:
        return noun + "s"


words = ["cat", "dog", "box", "bus", "brush"]

for word in words:
    print(word, "->", plural(word))
```

6.N-Gram Model for Text Generation (Bigram)

```python
import random
from collections import defaultdict
text = "I love machine learning. I love Python programming. I love machine learning."
words = text.lower().replace(".", "").split()
bigrams = defaultdict(list)
for i in range(len(words) - 1):
    bigrams[words[i]].append(words[i + 1])
current = "i"
generated = [current]
for i in range(10):
    if current not in bigrams:
        break
    current = random.choice(bigrams[current])
    generated.append(current)
print("Original Text:", text)
print("Generated Text:", " ".join(generated))
```

8.Stochastic Part-of-Speech Tagging

```python
from collections import defaultdict, Counter
training_data = [
    [("I", "PRON"), ("love", "VERB"), ("Python", "NOUN")],
    [("I", "PRON"), ("like", "VERB"), ("Python", "NOUN")],
    [("Students", "NOUN"), ("love", "VERB"), ("learning", "NOUN")],
    [("Python", "NOUN"), ("is", "VERB"), ("easy", "ADJ")],
    [("Students", "NOUN"), ("like", "VERB"), ("learning", "NOUN")]
]
word_tags = defaultdict(Counter)
for sentence in training_data:
    for word, tag in sentence:
        word_tags[word.lower()][tag] += 1
text = "I love Python"
print("STOCHASTIC POS TAGGING")
print("----------------------")

for word in text.split():
    key = word.lower()

    if key in word_tags:
        tag = word_tags[key].most_common(1)[0][0]
    else:
        tag = "UNKNOWN"

    print(word, "->", tag)
```

9.Rule-Based POS Tagging Using Regular Expressions

```python
import re
def pos_tag(word):
    word = word.lower()

    if re.fullmatch(r"\d+", word):
        return "NUMBER"

    elif re.fullmatch(r"\w+ing", word):
        return "VERB"

    elif re.fullmatch(r"\w+ly", word):
        return "ADVERB"

    elif re.fullmatch(r"\w+(ous|ful|able)", word):
        return "ADJECTIVE"

    elif re.fullmatch(r"\w+ed", word):
        return "VERB"

    elif re.fullmatch(r"\w+s", word):
        return "NOUN (PLURAL)"

    else:
        return "NOUN"
text = "Students are learning Python quickly"
print("RULE-BASED POS TAGGING")
print("----------------------")
for word in text.split():
    print(word, "->", pos_tag(word))
```
