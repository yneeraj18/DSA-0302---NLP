```python
words = ["connected", "connecting", "connection"]

print("{:<15} {:<15} {:<15} {:<20} {:<15}".format(
    "Word", "Root", "Suffix", "Suffix Type", "Normalized Form"
))
print("-" * 80)

for word in words:
    if word == "connected":
        root = "connect"
        suffix = "ed"
        suffix_type = "Inflectional"
        normalized = "connect"

    elif word == "connecting":
        root = "connect"
        suffix = "ing"
        suffix_type = "Inflectional"
        normalized = "connect"

    elif word == "connection":
        root = "connect"
        suffix = "ion"
        suffix_type = "Derivational"
        normalized = "connect"

    print("{:<15} {:<15} {:<15} {:<20} {:<15}".format(
        word, root, suffix, suffix_type, normalized
    ))
```

```python
words = ["unhappy", "happiness", "happily"]

print("{:<12} {:<12} {:<12} {:<15} {:<15} {}".format(
    "Word", "Prefix", "Base", "Suffix", "Type", "Normalized Root"
))
print("-" * 85)

for word in words:
    if word == "unhappy":
        prefix = "un-"
        base = "happy"
        suffix = "None"
        morph_type = "Derivational"
        root = "happy"

    elif word == "happiness":
        prefix = "None"
        base = "happy"
        suffix = "-ness"
        morph_type = "Derivational"
        root = "happy"

    elif word == "happily":
        prefix = "None"
        base = "happy"
        suffix = "-ly"
        morph_type = "Derivational"
        root = "happy"

    print("{:<12} {:<12} {:<12} {:<15} {:<15} {}".format(
        word, prefix, base, suffix, morph_type, root
    ))
```

```python
words = ["played", "player", "playing"]

print("{:<12} {:<12} {:<12} {:<18} {}".format(
    "Word", "Stem", "Removed Affix", "Transformation", "Normalized Form"
))
print("-" * 75)

for word in words:
    if word == "played":
        stem = "play"
        affix = "-ed"
        morph_type = "Inflectional"
        normalized = "play"

    elif word == "player":
        stem = "play"
        affix = "-er"
        morph_type = "Derivational"
        normalized = "play"

    elif word == "playing":
        stem = "play"
        affix = "-ing"
        morph_type = "Inflectional"
        normalized = "play"

    print("{:<12} {:<12} {:<12} {:<18} {}".format(
        word, stem, affix, morph_type, normalized
    ))
```

```python
words = ["writes", "writing", "written"]

morphology = {
    "writes": {
        "root": "write",
        "suffix": "-s",
        "type": "Regular inflection",
        "path": "START -> ROOT -> S -> ACCEPT"
    },
    "writing": {
        "root": "write",
        "suffix": "-ing",
        "type": "Regular inflection",
        "path": "START -> ROOT -> ING -> ACCEPT"
    },
    "written": {
        "root": "write",
        "suffix": "-en",
        "type": "Irregular inflection",
        "path": "START -> IRREGULAR FORM -> ACCEPT"
    }
}

print("{:<12} {:<10} {:<10} {:<22} {}".format(
    "Word", "Root", "Suffix", "Inflection Type", "State Transition Path"
))
print("-" * 105)

for word in words:
    data = morphology[word]

    print("{:<12} {:<10} {:<10} {:<22} {}".format(
        word,
        data["root"],
        data["suffix"],
        data["type"],
        data["path"]
    ))
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["relational", "relation", "relate"]

print("{:<15} {:<25} {:<20}".format(
    "Original Word", "Applied Rule", "Final Stem"
))
print("-" * 60)

for word in words:
    if word == "relational":
        rule = "Remove -ational"
    elif word == "relation":
        rule = "Remove -ion"
    else:
        rule = "Remove -e"

    stem = ps.stem(word)

    print("{:<15} {:<25} {:<20}".format(
        word, rule, stem
    ))
```
